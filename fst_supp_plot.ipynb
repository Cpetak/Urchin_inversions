```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
import re

file_path = Path.home() / "Desktop" / "new_fsts" / "NW_022145610.1_30779143_31460853_homop_hetero_fst.weir.fst"

df = pd.read_csv(file_path, sep="\t")

# extract numeric window from filename
m = re.search(r"_(\d+)_(\d+)_", file_path.name)
start, end = map(int, m.groups())

# add 1% padding
pad = int(0.1 * (end - start))
xmin, xmax = start - pad, end + pad

# plot
plt.figure(figsize=(10, 4))
plt.scatter(df["POS"], df["WEIR_AND_COCKERHAM_FST"], s=5)

plt.xlabel("Position (POS)")
plt.ylabel("Weir and Cockerham FST")
plt.title(file_path.name)

#plt.xlim(xmin, xmax)
plt.axhline(0, linewidth=1, color="black")

plt.axvline(xmin, color="red", linestyle="--", label="Start")
plt.axvline(xmax, color="red", linestyle="--", label="End")

plt.tight_layout()
plt.show()
```
